```python
# read a file
f = open("crappydb.csv")

line = f.readline()
while line != '':
    print(line)
    line = f.readline()
f.close()
```

```python
# better approach

f = open("crappydb.csv")
for line in f:
    print(line)
f.close()
```

```python
# better than open/close

# with statement
with open('crappydb.csv') as f:
    for line in f:
        print(line)
```

```python
# writing to a file
# overwrites current contents
with open('crappydb.csv', 'w') as f:
    f.write('W999000,red,rudolph\n')
```

```python
# appending to a file
# doesn't overwrite the current contents

with open('crappydb.csv', 'a') as f:
    f.write('W000111,santa,claus\n')
```

```python
# delete a line

# read contents
db = []
with open('crappydb.csv', 'r') as f:
    for line in f:
        db.append(line)

# remove unwanted content
db.pop()

# write contents
with open('crappydb.csv', 'w') as f:
    for r in db:
        f.write(r)
```
